```python
"""small loader script to test loading mimic_rrs dataset from disk"""

import datasets
import pprint
import pandas as pd
```

```python
ds = datasets.load_from_disk("datasets/medical/mimic_rrs")
```

```python
pprint.pprint(ds[0])
print(type(ds[0]))
print(ds[0].keys())
print(len(ds))

print(ds)
```

```python
df = pd.DataFrame(ds)

df
```

```python
num_fin_text = df[df["TEXT"].str.contains("FINDINGS", case=False, na=False)]
num_imp_text = df[df["TEXT"].str.contains("IMPRESSION", case=False, na=False)]

print(len(num_fin_text))
print(len(num_imp_text))

print("\n\n\n\n\n")

num_fin_report = df[df["REPORT"].str.contains("FINDINGS", case=False, na=False)]
num_imp_report = df[df["REPORT"].str.contains("IMPRESSION", case=False, na=False)]

print(len(num_fin_report))
print(len(num_imp_report))
```
